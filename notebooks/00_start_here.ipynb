# 00 — Start here: v0.2.0 publication experiment

This project freezes **v0.1.7** as the engineering baseline and starts a separate **v0.2.0 scientific experiment**. The rule is: train on `train`, tune/freeze on `validation`, and evaluate once on the 2,000-image `test` split. No test-derived parameter tuning is allowed.

Pipeline: `01 dataset → 02 baseline audit → 03 freeze net payloads → 04 auxiliary block-risk model → 05 tune/freeze allocator → 06 frozen test experiment → 07 independent steganalysis → 08 robustness/resources → 09 publication outputs`.

```python
from pathlib import Path
import yaml
from rdhlab import __version__
from rdhlab.environment import capture_environment

CFG=Path('/workspace/config/experiment.yaml')
config=yaml.safe_load(CFG.read_text())
run_dir=capture_environment('/workspace/results')
print('rdhlab version:', __version__)
print('baseline frozen at:', config['project']['baseline_version'])
print('current experiment:', config['project']['version'])
print('run directory:', run_dir)
print('seed:', config['project']['seed'])
```

```python
import pandas as pd
manifest_path=Path(config['dataset']['prepared_manifest'])
if manifest_path.exists():
    manifest=pd.read_csv(manifest_path)
    print('Manifest:', manifest_path)
    print(manifest['split'].value_counts())
    print('Images:', len(manifest), 'size:', manifest[['width','height']].drop_duplicates().to_dict('records'))
else:
    print('Dataset manifest is not present yet. Run 01_prepare_dataset.ipynb.')
```

**Reproducibility rule.** `config/frozen_payloads.json` and `config/frozen_allocator.json` are created only from the validation split. Once the frozen test experiment has started, do not overwrite these files; create a new experiment version instead.
